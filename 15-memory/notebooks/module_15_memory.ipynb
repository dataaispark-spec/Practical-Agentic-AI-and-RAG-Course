# Module 15 — Agent Memory Engineering
Framework-free lab: working, episodic, semantic and procedural memory.

```python
from dataclasses import dataclass
from datetime import datetime, timezone, timedelta
import re
@dataclass
class Memory:
    id:str; tenant:str; kind:str; content:str; confidence:float=1.0; importance:float=.5; expires_at:object=None
class Store:
    def __init__(self): self.items={}
    def put(self,m): self.items[m.id]=m
    def search(self,q,tenant,k=5):
        qw=set(re.findall(r'\w+',q.lower())); out=[]
        for m in self.items.values():
            if m.tenant!=tenant: continue
            if m.expires_at and m.expires_at<=datetime.now(timezone.utc): continue
            mw=set(re.findall(r'\w+',m.content.lower())); rel=len(qw&mw)/max(1,len(qw))
            if rel: out.append((rel*m.confidence*(.5+.5*m.importance),m))
        return [m for _,m in sorted(out,reverse=True,key=lambda x:x[0])[:k]]
s=Store(); s.put(Memory('1','A','semantic','Python async client uses bounded concurrency',.95,.9)); s.put(Memory('2','A','episodic','Yesterday the client timed out',.8,.5)); print(s.search('Python client','A'))
```
