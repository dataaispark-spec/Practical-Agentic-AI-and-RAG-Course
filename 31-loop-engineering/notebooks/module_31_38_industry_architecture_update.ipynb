# Cross-module enterprise scenarios

## 1. SOC autonomous investigation
Alert → enrichment → internal RAG → evidence correlation → tool calls → approval for containment → verification → incident record.

## 2. IT operations teammate
Ticket → service discovery → runbook retrieval → safe remediation → SLO verification → escalation/change record.

## 3. Banking operations
Case → policy/evidence retrieval → recommendation → independent verification → approval → ledger/CRM update → audit package.

## 4. Enterprise research
Goal → web/internal retrieval → evidence graph → synthesis → citation verification → scheduled follow-up → durable state.

## 5. Software engineering teammate
Issue → repository exploration → plan → edit → tests → security checks → evaluator → PR/commit → CI recovery.

## 6. Procurement / sales operations
Research → CRM/procurement updates → draft communication → approval before external effect → verify submission → follow-up schedule.

For every scenario, compare: synchronous vs durable; API tools vs computer use; single vs multi-agent; memory/no-memory; RAG/no-RAG; deterministic vs model verification; and baseline vs self-improving harness.


````python
# Reusable architecture scoring skeleton
from dataclasses import dataclass
@dataclass
class ArchitectureScore:
    success: float
    safety: float
    latency: float
    cost: float
    recovery: float
    complexity: float

def score(a):
    return (0.30*a.success + 0.25*a.safety + 0.15*a.recovery - 0.10*a.latency - 0.10*a.cost - 0.10*a.complexity)

baseline=ArchitectureScore(.80,.99,.40,.20,.70,.20)
frontier=ArchitectureScore(.88,.98,.55,.45,.90,.65)
print("baseline:",score(baseline))
print("frontier:",score(frontier))
````
