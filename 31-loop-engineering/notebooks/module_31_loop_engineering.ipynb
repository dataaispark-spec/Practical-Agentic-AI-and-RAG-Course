# Module 31 — Loop Engineering: Production Agent Loop Engine

Build the control loop beneath an agent framework: OBSERVE → DECIDE → ACT → VERIFY → RECOVER → CONTINUE/STOP. The notebook treats halting, budgets, retries and verification as engineering primitives.

## Learning goals
- separate model decisions from deterministic control;
- define explicit state transitions;
- implement hard step/token/tool/time budgets;
- detect repetition and non-progress;
- distinguish retryable from permanent failures;
- checkpoint and recover;
- verify outcomes independently;
- record trajectories for evaluation.

```python
from dataclasses import dataclass, field
from enum import Enum
class Phase(Enum): OBSERVE=1; DECIDE=2; ACT=3; VERIFY=4; RECOVER=5; STOP=6
@dataclass
class State:
    goal:str; phase:Phase=Phase.OBSERVE; steps:int=0; max_steps:int=8; history:list=field(default_factory=list); verified:bool=False
```

## Lab 1 — Minimal finite-state loop
Implement each transition explicitly. Prove that every execution reaches STOP or a hard budget boundary.

```python
s=State('find policy answer')
while s.steps < s.max_steps and not s.verified:
    s.history.append(s.phase.name); s.steps += 1
    if s.phase is Phase.OBSERVE: s.phase=Phase.DECIDE
    elif s.phase is Phase.DECIDE: s.phase=Phase.ACT
    elif s.phase is Phase.ACT: s.phase=Phase.VERIFY
    elif s.phase is Phase.VERIFY: s.verified=True; s.phase=Phase.STOP
print(s.history, s.phase)
```
