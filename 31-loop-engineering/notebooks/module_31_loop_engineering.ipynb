Build the control loop beneath an agent framework: OBSERVE → DECIDE → ACT → VERIFY → RECOVER → CONTINUE/STOP. The notebook treats halting, budgets, retries and verification as engineering primitives.

````python
from dataclasses import dataclass, field
from enum import Enum
class Phase(Enum): OBSERVE=1; DECIDE=2; ACT=3; VERIFY=4; RECOVER=5; STOP=6
@dataclass
class State:
    goal:str; phase:Phase=Phase.OBSERVE; steps:int=0; max_steps:int=8; history:list=field(default_factory=list); verified:bool=False
````

## Lab 1 — Minimal finite-state loop
Implement each transition explicitly. Prove that every execution reaches STOP or a hard budget boundary.


````python
s=State('find policy answer')
while s.steps < s.max_steps and not s.verified:
    s.history.append(s.phase.name); s.steps += 1
    if s.phase is Phase.OBSERVE: s.phase=Phase.DECIDE
    elif s.phase is Phase.DECIDE: s.phase=Phase.ACT
    elif s.phase is Phase.ACT: s.phase=Phase.VERIFY
    elif s.phase is Phase.VERIFY: s.verified=True; s.phase=Phase.STOP
print(s.history, s.phase)
````

# 2026 Industry Architecture Update

The production loop is now a **governed execution kernel**, not merely a prompt pattern.

```text
Goal/Contract → OBSERVE → DECIDE → POLICY → ACT → VERIFY
                    ↑                     ↓
                 RECOVER ← CHECKPOINT ← AUDIT
                    └── Budget / Deadline / Approval / Kill switch
```

**Real-world scenarios**
- SOC: alert → enrich → investigate → approval-gated containment → verify endpoint state.
- IT operations: ticket → runbook retrieval → remediation proposal → change-policy check → execute → verify SLO.
- Banking: policy/evidence retrieval → recommendation → regulated-action approval → ledger verification.
- Coding agents: inspect → edit → test → repair → independent evaluation → commit.

The model proposes; deterministic software decides whether the proposal may become an external effect.


## Architecture upgrades to implement

1. Task contract: goal, completion criteria, risk, tools, budget, deadline.
2. Action envelope: identity, tenant, tool, argument hash, risk, idempotency key.
3. Independent verifier: inspect external state, not model claims.
4. Progress detector: repetition, oscillation, no-progress.
5. Recovery taxonomy: retry, re-observe, re-plan, escalate, rollback, terminate.
6. Durable checkpoints before/after consequential effects.
7. Operator kill switch independent of model cooperation.
8. Replayable trajectory for debugging/evaluation.
