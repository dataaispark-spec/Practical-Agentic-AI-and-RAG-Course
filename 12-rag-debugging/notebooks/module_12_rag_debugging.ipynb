# Module 12 — RAG Debugging
Colab-ready failure investigation lab. Diagnose the first failing stage from traces before inspecting implementation.

```python
traces=[
 {'run':'r1','spans':[('query','ok'),('retrieval','ok'),('ranking','error')]},
 {'run':'r2','spans':[('query','ok'),('retrieval','error')]},
 {'run':'r3','spans':[('query','ok'),('retrieval','ok'),('generation','error')]}
]
def first_failure(spans):
    for stage,status in spans:
        if status=='error': return stage
    return None
for t in traces: print(t['run'],first_failure(t['spans']))
```

```python
def classify(stage):
    return {'query':'query failure','retrieval':'candidate generation failure','ranking':'ranking failure','filter':'authorization/metadata failure','generation':'generation failure'}.get(stage,'system failure')
for t in traces: print(t['run'],classify(first_failure(t['spans'])))
```
