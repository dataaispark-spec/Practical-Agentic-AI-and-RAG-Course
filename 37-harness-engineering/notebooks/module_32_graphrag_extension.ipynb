# Module 32 Extension — GraphRAG Benchmark

Compare graph retrieval and vector retrieval fairly. Do not assume GraphRAG wins.

```python
def rrf(graph_ids, vector_ids, k=60):
    scores={}
    for rank,x in enumerate(graph_ids,1): scores[x]=scores.get(x,0)+1/(k+rank)
    for rank,x in enumerate(vector_ids,1): scores[x]=scores.get(x,0)+1/(k+rank)
    return [x for x,_ in sorted(scores.items(), key=lambda p:(-p[1],p[0]))]

graph=['B','C','D']; vector=['D','B','E']
print('graph:',graph,'vector:',vector,'hybrid:',rrf(graph,vector))
```

## Experiment matrix

Run vector-only, graph-only, graph+vector and graph+vector+reranker+verifier. Measure Recall@K, MRR/nDCG, multi-hop accuracy, groundedness, citation correctness, p50/p95 latency, tokens and cost per verified task.
