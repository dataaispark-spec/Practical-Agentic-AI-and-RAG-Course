## Industry architecture
Trajectory / Incident Evidence → Candidate Memory / Skill → Provenance + Trust + Tenant + Risk → Benchmark + Security + Regression → Approval / Version / Rollback → Runtime Retrieval → New Experience.

Examples: support runbooks, SRE procedures, SOC investigation patterns, coding workflows and account preferences.
