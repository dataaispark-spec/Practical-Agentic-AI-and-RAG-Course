Practice message construction, structured outputs, model routing, fallbacks, and deterministic control around a probabilistic model.


```python
from dataclasses import dataclass
@dataclass(frozen=True)
class Model:
 name:str; quality:float; cost:float; latency_ms:int; healthy:bool=True
models=[Model('fast',.82,.2,200),Model('quality',.95,1.0,700),Model('private',.90,.8,500)]
def route(min_quality=.85,max_cost=1,max_latency=800,private=False):
    candidates=[m for m in models if m.healthy and m.quality>=min_quality and m.cost<=max_cost and m.latency_ms<=max_latency and (not private or m.name=='private')]
    return min(candidates,key=lambda m:(m.cost,m.latency_ms)) if candidates else None
print(route()); print(route(private=True))
```
