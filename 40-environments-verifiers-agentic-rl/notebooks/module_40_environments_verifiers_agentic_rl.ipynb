# Module 40 — Environments, Verifiers & Agentic RL

Build the interaction-and-evaluation layer for agents: task → environment → trajectory → independent verification → reward/gate → evaluation. The emphasis is on objective signals that can withstand reward hacking and model-judge error.


## Industry architecture
```text
Task Generator → Environment ↔ Agent/Harness
                        ↓
                    Trajectory
                        ↓
              Independent Verifier(s)
                  ↓           ↓
             Reward       Safety Gate
                  ↓           ↓
             Filtering → Evaluation → Training/Promotion
```
Examples: code execution tests, browser workflows, RAG citation checks, IT health remediation and financial reconciliation.


````python
class Environment:
    def reset(self, task): return {'task':task,'step':0,'changes':[]}
    def step(self, state, action):
        state['step'] += 1
        state['changes'].append(action)
        done = state['step'] >= 3 or action == 'complete'
        return state, float(action == 'complete'), done, {'valid': action != 'unsafe'}
env=Environment(); s=env.reset('fix service health'); print(env.step(s,'inspect'))
````
