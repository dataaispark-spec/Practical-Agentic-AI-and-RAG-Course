# Module 5 — Prompting + Evaluation
Practice prompt design as controlled experimentation and build a small evaluation harness.


```python
from dataclasses import dataclass
@dataclass
class Case: prompt:str; expected:str
cases=[Case('2+2','4'),Case('Capital of France','Paris'),Case('Say YES if safe','YES')]
def exact_score(pred,expected): return int(pred.strip().lower()==expected.strip().lower())
preds=['4','Paris','YES']; print([exact_score(p,c.expected) for p,c in zip(preds,cases)])
```

## Lab 1 — Prompt variables
Change one variable at a time: role, instructions, examples, output schema, context ordering. Record quality and failure modes.

## Lab 2 — Few-shot vs zero-shot
Create a balanced test set and compare accuracy. Do not tune the prompt on the test set.


```python
def evaluate(predictions,cases):
    scores=[exact_score(p,c.expected) for p,c in zip(predictions,cases)]
    return sum(scores)/len(scores)
print('baseline:',evaluate(preds,cases))
```
