Build a planner/executor/verifier system with approval gates and replanning.

```python
from dataclasses import dataclass
@dataclass(frozen=True)
class Step: id:str; objective:str; deps:tuple=(); risk:str='low'
def validate(steps):
    ids={s.id for s in steps}
    assert all(d in ids for s in steps for d in s.deps), 'missing dependency'
    graph={s.id:set(s.deps) for s in steps}; seen=set(); active=set()
    def dfs(n):
        if n in active: raise ValueError('cycle')
        if n in seen:return
        active.add(n); [dfs(d) for d in graph[n]]; active.remove(n); seen.add(n)
    [dfs(n) for n in graph]
steps=[Step('collect','collect evidence'),Step('analyze','analyze',('collect',)),Step('report','report',('analyze',))]
validate(steps); print('valid DAG')
```

```python
def execute(steps, approved=False):
    done=set()
    for s in steps:
        if not set(s.deps)<=done: raise RuntimeError('precondition')
        if s.risk=='high' and not approved: raise PermissionError('approval required')
        print('EXEC',s.id); done.add(s.id)
execute(steps)
```
