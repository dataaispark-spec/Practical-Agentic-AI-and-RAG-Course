## Critical distinction
A verifier checks whether an outcome satisfies an objective. A language-model judge can be useful, but high-stakes rewards should use deterministic or independently validated signals whenever possible.

```python
class Environment:
    def reset(self): return {'step':0}
    def step(self, action): return {'step':1}, 1.0, True, {'valid': action == 'solve'}

env=Environment(); obs=env.reset(); print(obs, env.step('solve'))
```
