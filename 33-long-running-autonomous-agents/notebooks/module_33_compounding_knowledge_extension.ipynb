# Module 33 Extension — Compounding Knowledge

Karpathy-inspired pattern: repeatedly turn new evidence into maintained knowledge rather than accumulating an uncurated transcript.

```python
knowledge={}
contradictions=[]

def upsert(key, value, source):
    if key in knowledge and knowledge[key]['value'] != value:
        contradictions.append((key, knowledge[key], {'value':value,'source':source}))
        return 'contradiction'
    if key in knowledge: return 'duplicate'
    knowledge[key]={'value':value,'source':source}
    return 'added'

print(upsert('A.status','active','policy-v1'))
print(upsert('A.status','active','policy-v1'))
print(upsert('A.status','closed','policy-v2'))
print('knowledge=',knowledge)
print('contradictions=',contradictions)
```
