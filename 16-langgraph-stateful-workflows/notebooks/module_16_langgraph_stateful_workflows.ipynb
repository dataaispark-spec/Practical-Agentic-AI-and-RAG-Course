# Module 16 — Stateful Agent Workflows
First build the graph mechanics from scratch; then optionally install/use LangGraph to map concepts to a production framework.

```python
from dataclasses import dataclass,field
from enum import Enum
class Status(Enum): RUNNING='running'; WAITING='waiting'; COMPLETED='completed'; FAILED='failed'
@dataclass
class State: data:dict=field(default_factory=dict); current:str='start'; history:list=field(default_factory=list); status:Status=Status.RUNNING
class Graph:
    def __init__(self,limit=20): self.nodes={}; self.edges={}; self.limit=limit
    def node(self,n,fn): self.nodes[n]=fn
    def edge(self,a,b,cond=lambda s:True): self.edges.setdefault(a,[]).append((cond,b))
    def run(self,s):
        for _ in range(self.limit):
            if s.current=='END': s.status=Status.COMPLETED; return s
            s.history.append(s.current); self.nodes[s.current](s)
            if s.status is Status.WAITING:return s
            choices=self.edges.get(s.current,[]); nxt=next((b for c,b in choices if c(s)),None)
            if nxt is None:s.status=Status.COMPLETED;return s
            s.current=nxt
        s.status=Status.FAILED;return s
g=Graph(); g.node('start',lambda s:s.data.update(ok=True)); g.node('work',lambda s:s.data.update(answer=42)); g.edge('start','work',lambda s:s.data['ok']); g.edge('work','END'); print(g.run(State()))
```
