# Module 36 — Recursive / Self-Improving Agents

Build a controlled improvement platform where an agent can propose changes to prompts, retrieval, tools, skills or harness logic, but promotion remains externally gated by evidence.


## Industry architecture
```text
Production Failures / Feedback
          ↓
Failure Mining → Improvement Hypothesis
          ↓
Candidate Artifact → Isolated Sandbox
          ↓
Fixed + Adversarial + Security Evaluation
          ↓
Quality / Safety / Cost / Latency Gates
          ↓
Shadow → Canary → Promote
          ↓
Monitor → Detect Regression → Rollback
```
Targets: RAG quality, coding skills, support routing, tool selection and browser workflows.


````python
baseline={'success':0.80,'grounded':0.90,'safety':1.00,'cost':0.10}
candidate={'success':0.84,'grounded':0.92,'safety':1.00,'cost':0.11}
promote = candidate['success']>=baseline['success']+0.02 and candidate['safety']>=baseline['safety'] and candidate['cost']<=0.15
print('promotion gate:',promote)
````

## Lab 1 — Improve one measurable failure
Start from a concrete incident: enterprise RAG answers miss the authoritative policy because retrieval ranks an outdated document first. Define an improvement objective and non-negotiable safety conditions.


## Lab 2 — Freeze the baseline
Record model/provider, prompt, retrieval configuration, embedding version, reranker, tool versions, policies, dataset hash and evaluation protocol. Make the baseline reproducible.


## Lab 3 — Generate candidate improvements
Propose alternatives such as metadata-aware reranking, query rewriting, skill routing or context compaction. Store each as a versioned artifact rather than editing production directly.


## Lab 4 — Isolated experiment runner
Run baseline and candidate over identical fixtures. Capture quality, groundedness, citation correctness, safety, latency and cost. Prevent network or production-tool access during experimentation.


## Lab 5 — Statistical and policy gates
Require minimum quality gain, zero critical security regressions, bounded latency/cost and no forbidden behavior. Add paired-case analysis rather than relying on one average.


## Lab 6 — RAG self-improvement case
Compare: dense-only retrieval vs hybrid retrieval vs hybrid+rerank on 100 enterprise cases. Show how a candidate can improve factual accuracy while increasing latency, and choose using a multi-objective gate.


## Lab 7 — Coding-agent self-improvement
Mine failed repair trajectories for a new debugging skill. Validate it on unseen repositories, syntax failures, security tests and regression tests before promotion.


## Lab 8 — Benchmark contamination
Give a candidate access to evaluation metadata and demonstrate inflated results. Remove the leakage, regenerate identifiers and enforce a strict train/validation/test boundary.


## Lab 9 — Reward/metric gaming
Construct a candidate that increases citation count by adding irrelevant citations, or closes more support tickets without actually resolving them. Strengthen the evaluator to measure real task outcomes.


## Lab 10 — Change provenance
Record proposal origin, evidence set, experiment seed, artifact hash, evaluator version, approver, deployment window and rollback target. Reconstruct why the change was promoted.


## Lab 11 — Shadow and canary
Run the candidate in shadow mode, then route a small traffic slice to it. Define automatic rollback on safety regression, p95 latency breach or task-success degradation.


## Lab 12 — Post-promotion drift
Simulate corpus drift, tool API changes and user-behavior changes. Detect degradation and send the candidate back to evaluation/review.


## Lab 13 — Pareto frontier
Compare ten candidate improvements across quality, safety, latency and cost. Identify dominated candidates and choose a practical Pareto-optimal version.


## Lab 14 — Recursive harness improvement
Allow the improvement engine to propose a change to its own prompt/skill selection/evaluation harness. The same external gate must approve it; recursion depth and experiment budget are bounded.


## Lab 15 — Full improvement incident
Inject evaluator bug, benchmark leakage, candidate regression, safety degradation, runaway experiments and rollback failure. Build a recovery plan that preserves the known-good artifact.


## Interview/System Design challenge
Design a self-improving enterprise RAG platform for 1 million monthly requests. Explain experiment isolation, evaluation independence, canary policy, auditability, cost controls and rollback.


## Mastery gate
No candidate may reach production merely because it scores better on one metric. Promotion requires reproducible evidence across quality, safety, operations and governance.
