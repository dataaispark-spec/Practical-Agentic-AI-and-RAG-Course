Build a research-grade improvement loop: agent proposes a change, evaluates evidence, compares against a baseline, promotes only if gates pass, and retains rollback capability.

## Safe improvement loop
Goal → baseline → propose → experiment → verify → compare → gate → version → deploy → monitor → rollback.

```python
baseline={'success':.80,'cost':.10,'safety':1.0}
candidate={'success':.84,'cost':.11,'safety':1.0}
promote=candidate['success']>baseline['success'] and candidate['safety']>=baseline['safety']
print('promote:',promote)
```
