# Module 32 — Harness Engineering: AegisAI Agent Harness

Build the runtime that connects a model to context, tools, state, policy, budgets, verification and an execution environment. Treat the model as replaceable intelligence and the harness as the engineered control plane.

## Industry architecture
```text
Task Contract → Identity/Tenant → Context Builder
                         ↓
                    Model Adapter
                         ↓
                  Action Proposal
                         ↓
              Policy + Budget + Approval
                         ↓
               Tool/MCP Gateway
                         ↓
                  Environment/API
                         ↓
       Verify → Observe → State/Checkpoint → Trace
```
Real-world targets: coding harnesses, SRE, SOC, enterprise research and customer support.


````python
from dataclasses import dataclass, field
@dataclass
class HarnessState:
    tenant:str; task:str; messages:list=field(default_factory=list); observations:list=field(default_factory=list); tool_results:list=field(default_factory=list); steps:int=0; tokens:int=0; cost:float=0.0
state=HarnessState('tenant-a','investigate failed deployment')
print(state)
````

## Lab 1 — Provider-neutral model adapter
Implement one interface for two fake providers with different response formats. Normalize model, latency, token usage, stop reason and tool proposals.
**Failure:** provider returns malformed tool arguments.
**Interview:** why should provider-specific policy not leak into business logic?


## Lab 2 — Enterprise context builder
Construct context from task contract, system policy, tenant-scoped RAG evidence, trusted memory, recent observations and tool results. Assign priority tiers and enforce a context budget.
**Scenario:** an IT incident with 30 log chunks plus a critical security policy. Prove the policy remains present after truncation.


## Lab 3 — Tool gateway
Create typed tools for search_logs, get_service_health, create_ticket and request_change. Implement schema validation, authorization, idempotency and structured errors.
**Attack:** model attempts an admin-only action with a support role.


## Lab 4 — Harness state model
Separate ephemeral prompt context from durable state. Persist task identity, approved actions, checkpoints, tool effects and verifier results.
**Exercise:** restart after the tool call but before verification; demonstrate safe resume.


## Lab 5 — Policy and budget hooks
Build pre-action policy and post-action verification hooks that cannot be overridden by model output, retrieval text or memory. Add step, token, tool and monetary budgets.


## Lab 6 — Context management
Compare naive transcript growth, rolling summary and retrieval-refresh strategies. Measure answer quality, context tokens, latency and cost.
**Industry case:** a 2-hour SOC investigation where only the latest evidence and stable incident facts should remain active.


## Lab 7 — Recovery and provider fallback
Inject provider timeout, malformed response, tool outage and process restart. Define retry/fallback policy by error class and preserve state across fallback.


## Lab 8 — Skills and extension surface
Expose skills through versioned manifests containing capability, preconditions, required tools, risk and evaluation version. Compare safe skill registration with arbitrary prompt-injection-based 'plugins'.


## Lab 9 — Harness benchmarking
Build three harness variants: minimal, guarded and durable. Run the same enterprise cases and report task success, tool-call precision, latency, cost and safety violations.


## Lab 10 — Red-team harness
Inject prompt injection in a retrieved document, malicious tool output, oversized context, hidden instruction in memory and tenant-mismatch state. Demonstrate where each attack is blocked.


## Lab 11 — Deterministic replay
Freeze model/tool fixtures and record trajectory inputs/outputs. Replay the same incident after changing harness logic and classify behavioral differences.


## Lab 12 — Production hardening
Add cancellation, deadlines, concurrency limits, state versioning, rate limits, circuit breakers, audit trails, secret isolation and graceful shutdown.


## Lab 13 — Coding-agent harness
Scenario: fix a failing repository test. Restrict filesystem access to repository workspace, require tests to pass, cap patch attempts and reject network access unless explicitly authorized.


## Lab 14 — Customer-support harness
Combine CRM context, policy/RAG, order lookup and refund tool. Allow read-only actions automatically; require exact approval for refunds above a threshold. Verify the resulting CRM state.


## Lab 15 — Architecture tradeoff review
Compare framework-heavy versus framework-light harnesses. Defend what belongs in deterministic code, what may remain model-controlled and where the trust boundaries sit.


## Failure-first exercises
Break context ordering, tool schema validation, state persistence, policy hooks, provider fallback, cancellation, tenant isolation, checkpoint recovery, audit recording and budget enforcement. For every failure identify the first broken invariant.


## Gold challenge
Build AegisAI Harness v1 as a reusable runtime. Swap models and tools without rewriting the safety/control plane. Demonstrate replay, recovery, policy isolation, bounded context, measurable cost and end-to-end auditability.
