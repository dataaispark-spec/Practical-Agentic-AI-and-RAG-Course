## Harness anatomy
Model adapter + system context + tool gateway + loop + state/checkpoints + policy + budget + verification + observability + recovery.

```python
from dataclasses import dataclass, field
@dataclass
class HarnessState:
    messages:list=field(default_factory=list); tool_results:list=field(default_factory=list); steps:int=0; budget:int=10
state=HarnessState()
```
