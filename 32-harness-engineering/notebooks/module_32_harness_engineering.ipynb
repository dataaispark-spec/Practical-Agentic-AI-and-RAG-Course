# 2026 Industry Architecture Update

The harness should be treated as a **stable control plane around replaceable models**.

```text
Task → Identity → Context/Memory/RAG → Model Adapter
                                  ↓
                           Action Proposal
                                  ↓
                     Policy + Budget + Approval
                                  ↓
                         Tool/MCP Gateway
                                  ↓
                       External Environment
                                  ↓
                    Verification + Audit + Trace
                                  ↓
                        Durable State/Recovery
```

**Real-world examples**
- Coding: repo + shell + tests + git + evaluator.
- IT ops: ticket + CMDB + observability + runbooks + change approval.
- Support: CRM + knowledge base + order/refund APIs + escalation.
- Research: browser/search + internal RAG + citation verifier.
- Persistent teammate: scheduler + browser/desktop + memory + approval.

Anthropic's 2026 work is a useful reference for long-running harness design; the key course lesson is to reproduce the control mechanisms, not merely copy a vendor product.
