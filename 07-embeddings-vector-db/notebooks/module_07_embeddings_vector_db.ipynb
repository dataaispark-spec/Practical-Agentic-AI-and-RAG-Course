# Module 7 — Embeddings + Vector Databases
Understand vector geometry, similarity search, metadata filters, recall, ANN trade-offs, and versioning.


```python
import numpy as np
items=['refund policy','password reset','invoice payment','manager approval']
rng=np.random.default_rng(7); X=rng.normal(size=(len(items),8)); X/=np.linalg.norm(X,axis=1,keepdims=True)
q=X[0]+.05*rng.normal(size=8); q/=np.linalg.norm(q)
scores=X@q
for i in np.argsort(-scores): print(round(float(scores[i]),3),items[i])
```

## Lab 1 — Geometry
Normalize vectors, compare cosine similarity vs dot product, and show why normalization changes ranking.

## Lab 2 — Recall@K
Create a gold relevant set and calculate Recall@1/3. Vary embedding noise and observe retrieval degradation.


```python
gold={'refund policy','manager approval'}
top=[items[i] for i in np.argsort(-scores)[:3]]
print('Recall@3=',len(gold&set(top))/len(gold),top)
```
