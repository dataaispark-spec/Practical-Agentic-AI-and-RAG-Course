# Module 6 — RAG From First Principles
Build a tiny RAG engine without a framework: documents → chunks → embeddings → similarity → evidence → grounded answer.


```python
import re, math
docs=[('d1','Refunds are allowed within 30 days.'),('d2','Enterprise refunds require manager approval.'),('d3','Password resets require identity verification.')]
def embed(s):
    v=[0]*32
    for w in re.findall(r'\w+',s.lower()): v[hash(w)%32]+=1
    n=math.sqrt(sum(x*x for x in v)) or 1
    return [x/n for x in v]
def sim(a,b): return sum(x*y for x,y in zip(embed(a),embed(b)))
def retrieve(q,k=2): return sorted(docs,key=lambda d:sim(q,d[1]),reverse=True)[:k]
print(retrieve('How long do I have to request a refund?'))
```
