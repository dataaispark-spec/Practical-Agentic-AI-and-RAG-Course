Integrate the two graphs: knowledge graph for evidence + task graph/harness for controlled execution.

```python
knowledge_edges=[('A','OWNS','B','claim-1'),('B','SUPPORTS','C','claim-2')]
task_states=['OBSERVE','DECIDE','ACT','VERIFY']

def graph_context(start, max_hops=2):
    assert max_hops <= 2, 'cap graph context'
    return [e for e in knowledge_edges if e[0]==start or e[2]==start]

print('knowledge evidence:', graph_context('A'))
print('task graph:', ' -> '.join(task_states))
```
